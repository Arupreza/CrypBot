```python
def scanner():
    active_symbols = Support_Resistance.get_active_symbols()
    btc_out = only_chandelier("BTCUSDT")
    btc_out = btc_out[198:]
    out_buy = int(btc_out['buy_signal'].sum())
    BTC_Support_Resistance = pd.DataFrame(Support_Resistance.Support_Resistance_P(['BTCUSDT']))


    if out_buy == 2:
        if BTC_Support_Resistance['near_resistance'][0] != True:
            print("Long Signal...")
            results_long = Chandelier_ZLSMA_Scan(
                coin_list=active_symbols,
                signal_type='buy'
            )
            alts_long =  pd.DataFrame(Support_Resistance.Support_Resistance_P(results_long['symbol'].tolist()))
            alts_long = alts_long[alts_long['signal'] != 'RESISTANCE']
            results_long = Chandelier_ZLSMA_Scan(
                coin_list=alts_long['symbol'].tolist(),
                signal_type='buy'
            )
        
    else:
        if BTC_Support_Resistance['near_support'][0] != True:
            print("Short Signal...")
            results_short = Chandelier_ZLSMA_Scan(
                coin_list=active_symbols,
                signal_type='sell'
            )
            alts_short =  pd.DataFrame(Support_Resistance.Support_Resistance_P(results_short['symbol'].tolist()))
            alts_short = alts_short[(alts_short['signal'] == 'RESISTANCE') | (alts_short['signal'] == 'NEUTRAL')]
            results_short = Chandelier_ZLSMA_Scan(
                coin_list=alts_short['symbol'].tolist(),
                signal_type='sell'
            )
            
    return alts_long if 'alts_long' in locals() else alts_short if 'alts_short' in locals() else None
```

```python
out = scanner()
```
